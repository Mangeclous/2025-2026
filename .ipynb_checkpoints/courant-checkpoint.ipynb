## Vérification des dépenses
- Ajout d'une colonne "Justificatif ?" pour la présence d'une pièce justificative correspondant à la dépense (valeur initiale : False)
- Ajout d'une colonne "Commentaires" pour toute note complémentaire utile.

### Fonctions auxiliaires - Vérification des dépenses
    

```python
def check_ok_montant(montant, commentaire, ok = True, df = dp):
  to_update = df['Montant'] == montant
  df.loc[to_update, 'Justificatif ?'] = ok
  df.loc[to_update, 'Commentaires'] = commentaire

def check_ok_contrepartie(contrepartie, commentaire = "Facture présente", ok = True, df = dp):
  df.loc[df["Contrepartie"] == contrepartie, "Commentaires"] = commentaire
  df.loc[df["Contrepartie"] == contrepartie, "Justificatif ?"] = ok

def check_ok_contreparties(*contreparties):
  for contrepartie in contreparties :
    check_ok_contrepartie(contrepartie)
```
